```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

prompt = "i am a unity expert and i wrote three aarticles about  "

result = generator(
    prompt,
    max_new_tokens=40,
    num_return_sequences=1,
    truncation=True
)

print(result[0]["generated_text"])
```

```python
for temp in [0.2,0.7,1.3]:
    result = generator(
        prompt, max_new_tokens=40, do_sample=True,
        temperature=temp , truncation=True
    )

    print(f"temperature{temp}:")
    print(result[0]["generated_text"])
    print("---")
```

```python

# STEP 1: LOAD & INSPECT

prompts_path = "/kaggle/input/datasets/javeriaismail99/writing-prompts/writingPrompts/train.wp_source"

with open(prompts_path, "r", encoding="utf-8") as f:
    all_prompts = f.readlines()

# Select the first five real writing prompts
sample_prompts = [p.strip() for p in all_prompts[:5]]

# Display the prompts
for i, p in enumerate(sample_prompts):
    print(f"{i+1}. {p}")
    print()
```

```python

# STEP 2: LOAD A PRETRAINED MODEL

from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

print("GPT-2 model loaded successfully!")
```

```python

# STEP 3: GENERATE STORY COMPLETIONS

completions = []

for p in sample_prompts:
    result = generator(
        p,
        max_new_tokens=60,
        num_return_sequences=1,
        truncation=True
    )
    completions.append(result[0]["generated_text"])

# Display all five generated completions
for i, c in enumerate(completions):
    print(f"COMPLETION {i+1}:")
    print(c)
    print("-" * 80)
```
